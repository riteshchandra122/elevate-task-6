```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
from sklearn.pipeline import Pipeline

# Load Iris
df = pd.read_csv("Iris.csv")

# Drop ID-like column if present
id_cols = [c for c in df.columns if c.lower() in ("id",)]
if id_cols:
    df = df.drop(columns=id_cols)

# Features/target
feature_cols = [c for c in df.columns if c.lower() != "species"]
X = df[feature_cols].values
y_raw = df["Species"].values

# Encode labels
le = LabelEncoder()
y = le.fit_transform(y_raw)

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

# Baseline pipeline: scaler + KNN(k=5)
pipe_knn = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])
pipe_knn.fit(X_train, y_train)

y_pred = pipe_knn.predict(X_test)
print(f"[Baseline KNN] k=5 — Test Accuracy: {accuracy_score(y_test, y_pred):.3f}")
print("\nClassification Report (Baseline):")
print(classification_report(y_test, y_pred, target_names=le.classes_))

cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm, display_labels=le.classes_).plot(values_format="d")
plt.title("Confusion Matrix — Baseline KNN (k=5)")
plt.show()
```

```python
k_values = range(1, 26)
train_accs, test_accs = [], []

for k in k_values:
    pipe = Pipeline([("scaler", StandardScaler()),
                     ("knn", KNeighborsClassifier(n_neighbors=k))])
    pipe.fit(X_train, y_train)
    train_accs.append(accuracy_score(y_train, pipe.predict(X_train)))
    test_accs.append(accuracy_score(y_test,  pipe.predict(X_test)))

plt.figure(figsize=(7,4))
plt.plot(list(k_values), train_accs, marker="o", label="Train")
plt.plot(list(k_values), test_accs, marker="s", label="Test")
plt.xlabel("k (n_neighbors)")
plt.ylabel("Accuracy")
plt.title("K vs Accuracy (Scaled features)")
plt.grid(alpha=0.3)
plt.legend()
plt.show()

best_k = list(k_values)[int(np.argmax(test_accs))]
print(f"Best k on held-out test: {best_k} (Test Accuracy={max(test_accs):.3f})")
```

```python
best_pipe = Pipeline([("scaler", StandardScaler()),
                      ("knn", KNeighborsClassifier(n_neighbors=best_k))])
best_pipe.fit(X_train, y_train)

y_best = best_pipe.predict(X_test)
print(f"[Refit Best] k={best_k} — Test Accuracy: {accuracy_score(y_test, y_best):.3f}")

cm_best = confusion_matrix(y_test, y_best)
ConfusionMatrixDisplay(cm_best, display_labels=le.classes_).plot(values_format="d")
plt.title(f"Confusion Matrix — Best KNN (k={best_k})")
plt.show()
```

```python
from sklearn.decomposition import PCA

# Pipeline for 2D visualization: scaler -> PCA(2) -> KNN(best_k)
from sklearn.pipeline import Pipeline
pca2_knn = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=2, random_state=42)),
    ("knn", KNeighborsClassifier(n_neighbors=best_k))
])
pca2_knn.fit(X_train, y_train)

scaler = pca2_knn.named_steps["scaler"]
pca    = pca2_knn.named_steps["pca"]
knn2d  = pca2_knn.named_steps["knn"]

X_train_pca = pca.transform(scaler.transform(X_train))
X_test_pca  = pca.transform(scaler.transform(X_test))

# Mesh in PCA space
x_min, x_max = X_train_pca[:,0].min()-1, X_train_pca[:,0].max()+1
y_min, y_max = X_train_pca[:,1].min()-1, X_train_pca[:,1].max()+1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 400),
                     np.linspace(y_min, y_max, 400))
grid_pca = np.c_[xx.ravel(), yy.ravel()]
Z = knn2d.predict(grid_pca).reshape(xx.shape)

# Plot (train overlay)
plt.figure(figsize=(7,6))
plt.contourf(xx, yy, Z, alpha=0.25)
plt.scatter(X_train_pca[:,0], X_train_pca[:,1], c=y_train, edgecolor="k", s=25)
plt.title(f"KNN Decision Boundary in PCA(2D) — Train (k={best_k})")
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.tight_layout(); plt.show()

# Plot (test overlay)
plt.figure(figsize=(7,6))
plt.contourf(xx, yy, Z, alpha=0.25)
plt.scatter(X_test_pca[:,0], X_test_pca[:,1], c=y_test, edgecolor="k", s=25)
plt.title(f"KNN Decision Boundary in PCA(2D) — Test (k={best_k})")
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.tight_layout(); plt.show()
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(best_pipe, X, y, cv=cv, scoring="accuracy")
print(f"Best KNN (scaled) — 5-fold CV Accuracy: mean={cv_scores.mean():.3f}, std={cv_scores.std():.3f}")
```
